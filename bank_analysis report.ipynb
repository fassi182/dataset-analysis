# Bank Marketing Campaign Success Prediction - Report

## Objective
To predict whether a customer will subscribe to a term deposit using the Bank Marketing Dataset. The task involves data preprocessing, feature selection, model training (Logistic Regression, Decision Tree, Random Forest), and performance evaluation.

---

## Dataset Overview
- **Source**: Bank Marketing Dataset
- **Target Variable**: `deposit` (yes/no)
- **Features**: Client demographics, financial data, and campaign-related attributes

---

## Preprocessing
- Replaced `"unknown"` values with `NaN` and filled missing values using the most frequent category.
- Categorical variables were encoded using one-hot encoding.
- All features were standardized using `StandardScaler`.

---

## Feature Selection
Top 10 features selected using mutual information:

- `poutcome_success`
- `housing_yes`
- `month_may`
- `contact_telephone`
- `balance_29`
- `balance_243`
- `balance_756`
- `balance_708`
- `duration_1047`
- `pdays_532`

---

## Models Used
| Model               | Accuracy | Precision | Recall | F1-score |
|---------------------|----------|-----------|--------|----------|
| Logistic Regression | 0.6233   | 0.5938    | 0.6489 | 0.6201   |
| Decision Tree       | 0.6235   | 0.5940    | 0.6495 | 0.6205   |
| Random Forest       | 0.6235   | 0.5940    | 0.6495 | 0.6205   |

> Note: All models performed similarly with moderate accuracy. Cross-validation was used to avoid overfitting.

## Conclusion
- All three models performed with nearly identical metrics (~62% accuracy).
- `poutcome_success` (previous campaign outcome) was by far the most influential feature.
- Followed by `housing_yes` and `month_may`.
- Other features had minimal impact, suggesting that past campaign success is the key driver of current subscription prediction.
