## Bank Marketing Dataset Summary

- **Dataset Shape:** `(11162, 17)`
- **Source:** Direct marketing campaigns of a Portuguese bank.

### Basic Overview

This dataset contains data related to a bank's marketing campaign aimed at persuading clients to subscribe to a **term deposit**. Each row represents a client, with various attributes describing their personal and contact details, as well as their interaction history with previous marketing efforts.

---

### Features Breakdown

#### 1. **Client Demographic Information**
- `age`: Age of the client.
- `job`: Type of job (e.g., admin., technician, services).
- `marital`: Marital status (e.g., married, single).
- `education`: Education level.
- `default`: Whether the client has credit in default.
- `balance`: Account balance.
- `housing`: Has a housing loan? (yes/no).
- `loan`: Has a personal loan? (yes/no).

#### 2. **Contact Information**
- `contact`: Communication type used (cellular/telephone).
- `day`: Last contact day of the month.
- `month`: Last contact month.
- `duration`: Duration of the last contact (in seconds).

#### 3. **Campaign History**
- `campaign`: Number of contacts performed during this campaign.
- `pdays`: Days since the client was last contacted (999 means never).
- `previous`: Number of contacts before this campaign.
- `poutcome`: Outcome of the previous campaign.

#### 4. **Target Variable**
- `deposit`: Whether the client subscribed to a term deposit (`yes` or `no`).

---

### Use Case

This dataset is commonly used for:
- **Classification problems**
- **Customer behavior analysis**
- **Predictive modeling in marketing**

The objective is to build a model to predict whether a client will subscribe to a term deposit based on their profile and interaction history.

### Basic Overview

This dataset combines **on-court performance metrics** of NBA players from the **2016-2017 season** with their **social media influence** data. Each row represents an individual player, capturing their performance statistics, social media engagement, and salary information.

---

### Features Breakdown

#### 1. **Player Performance Metrics**
- `points_per_game`: Average points scored per game.
- `assists_per_game`: Average assists made per game.
- `rebounds_per_game`: Average rebounds per game.
- `steals_per_game`: Average steals per game.
- `blocks_per_game`: Average blocks per game.
- *Additional performance metrics as available.*

#### 2. **Social Media Influence**
- `twitter_followers`: Number of Twitter followers.
- `twitter_engagement`: Engagement metrics on Twitter (likes, retweets, etc.).
- `wikipedia_pageviews`: Number of views on the player's Wikipedia page.

#### 3. **Salary Information**
- `salary`: Player's salary for the 2016-2017 season.

---

### Use Case

This dataset is valuable for:
- **Analyzing the correlation** between a player's on-court performance and their social media influence.
- **Predicting factors** that contribute to a player's popularity and marketability.
- **Exploring the impact** of social media presence on salary and endorsements.

Researchers and analysts can utilize this dataset to build models that assess how performance metrics and social influence intertwine in the realm of professional basketball.
